What is the difference between hardware and software? Give one example of each.
Hardware : For a computer, hardware is electronic devices that are connected to control computer operations, input and output operations.
  Examples : CPU, keyboard, mouse etc.,
Software : A set of instructions given to computer which is called program and later it gets converted into machine understandable language(binary language).
 Examples : Windows OS or any other OS is System Software
            Microsoft Office is Application Software.
